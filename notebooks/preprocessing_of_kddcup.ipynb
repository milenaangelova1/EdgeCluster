```python
df = df[['duration',
'src_bytes',
'dst_bytes',
'wrong_fragment',
'urgent',
'hot',
'num_failed_logins',
'num_compromised',
'root_shell',
'su_attempted',
'num_root',
'num_file_creations',
'num_shells',
'num_access_files',
'num_outbound_cmds',
'count',
'srv_count',
'serror_rate',
'srv_serror_rate',
'rerror_rate',
'srv_rerror_rate',
'same_srv_rate',
'diff_srv_rate',
'srv_diff_host_rate',
'dst_host_count',
'dst_host_srv_count',
'dst_host_same_srv_rate',
'dst_host_diff_srv_rate',
'dst_host_same_src_port_rate',
'dst_host_srv_diff_host_rate',
'dst_host_serror_rate',
'dst_host_srv_serror_rate',
'dst_host_rerror_rate',
'dst_host_srv_rerror_rate','target']]
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.fit(df['target'].unique())
target = le.transform(df['target'])
```

```python
df['target'] = target
```

```python
df
```

```python
df['timestamps']= list(range(1, df.shape[0]+1))
```

```python
scaler = MinMaxScaler()
target = df['target']
timestamps = df['timestamps']
columns = df.columns
df = pd.DataFrame(scaler.fit_transform(df.drop(['target', 'timestamps'], axis=1).values), columns=columns[0:-2])
df
```

```python
df['timestamps'] = timestamps
df['cluster'] = target
```

```python
df.to_csv('../data/kddcup/preprocessed/kddcup.csv', index=False)
```

```python
path = '../data/kddcup/preprocessed'

df_split = np.array_split(df, 12)

for i, part in enumerate(df_split):
    part.to_csv(os.path.join(path, f'kddcup_{i}.csv'), index=False)
```
