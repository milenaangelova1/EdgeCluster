```python
import pandas as pd
import numpy as np
import os
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('../data/covertype/original/covtype.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
df['Cover_Type'] = df['Cover_Type'].astype(int)
```

```python
len(df.columns)
```

```python
df.drop(df.columns[10:-1], axis=1, inplace=True)
```

```python
df
```

```python
df['timestamps'] = list(range(1, df.shape[0]+1))
```

```python
df
```

```python
scaler = MinMaxScaler()
target = df['Cover_Type']
timestamps = df['timestamps']
columns = df.columns
df = pd.DataFrame(scaler.fit_transform(df.drop(['Cover_Type', 'timestamps'], axis=1).values), columns=columns[0:-2])
df
```

```python
df['timestamps'] = timestamps
df['cluster'] = target
```

```python
path = '../data/covertype/preprocessed'

df_split = np.array_split(df, 50)

for i, part in enumerate(df_split):
    part.to_csv(os.path.join(path, f'covertype_{i}.csv'), index=False)
```
