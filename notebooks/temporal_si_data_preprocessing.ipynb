```python
import pandas as pd
import os
import numpy as np
```

```python
base_dir = '../data/dataS/sequential-p/original'

dfs = []
files = next(os.walk(base_dir))[2]

for file in sorted(files)[1:]:
    dfs.append(pd.read_csv(os.path.join(base_dir, file), encoding='utf-8'))
```

```python
df = pd.read_csv('/Users/milenaangelova/git-repo/EdgeCluster/data/dataR/retail.arff.csv')
dfs = [df]
```

```python
# path = '../data/dataS/sequential-p/preprocessed'
path = '../data/dataR/retail/preprocessed'
for index, df in enumerate(dfs):
    df_split = np.array_split(df, 5)

    for i, part in enumerate(df_split):
        part.rename(columns={'class': 'cluster'}, inplace=True)
        part.to_csv(os.path.join(path, f'retail_{index}_{i}.csv'), index=False)
```
