```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
base_dir = '../data/synthetic/8-dim/'

dfs = []
files = next(os.walk(base_dir))[2]

for file in files:
    dfs.append(pd.read_csv(os.path.join(base_dir, file)))
```

```python
df1 = dfs[0]
df1.head()
```

```python
df1.info()
```

```python
df1.plot.box()
```

```python
df1.cluster.value_counts()
```

```python
df1['cluster'].unique()
```

```python
scatter = plt.scatter(df1["0"], df1["1"], df1["2"], df1["3"], df1["4"], df1["5"], df1["6"], df1["7"], c=df1['cluster'], alpha = 0.6, s=20)
plt.legend(handles=scatter.legend_elements()[0], labels=map(lambda x: 'Cluster ' + str(x), df1['cluster'].unique()))
```

```python
for index, df in enumerate(dfs):
    fig, ax = plt.subplots()
    ax.scatter(df["0"], df["1"], c=df['cluster'].unique(), alpha = 0.6, s=20)
    ax.legend(handles=scatter.legend_elements()[0], labels=map(lambda x: 'Cluster ' + str(x), df['cluster'].unique()))
    ax.set_title(f"Dataset {index}")
```
